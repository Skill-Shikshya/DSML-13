## Reading Data From Files

- `pd.read_csv(file_path, delimiter=',')`: Reads CSV (Comma Separated Value) file and returns a Pandas DataFrame
    - *file_path*: Path to csv file
    - *delimiter*: Symbol separating different columns (default=',')

```python
titanic_df = pd.read_csv("titanic_data.csv", delimiter=",", header = 0)
```

```python
titanic_df.shape
```

- `pd.read_json(file_path, orient="records")`: Reads Json file and returns a Pandas DataFrame
    - *file_path*: Path to csv file
    - *orient*: Indication of expected JSON string format

```python
json_df = pd.read_json("users.json")
```

```python
type(json_df)
```
