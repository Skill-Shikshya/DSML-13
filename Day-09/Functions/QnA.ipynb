```python
number = 18

for i in range(2, number):
    if number % i == 0:
        print(f"{number} is not a prime number.")
        break
else:
    print(f"{number} is a prime number.")
```

### The Nested Comprehension Trap
#### Topic: Nested List Comprehension

Write a single-line list comprehension that flattens a 2D matrix, but only includes even numbers, and squares them.


```python
# Python
# Input matrix:
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Expected Output: [4, 16, 36, 64]
# Your code here:
squared_evens = [ num ** 2 for items in matrix for num in items if num % 2 == 0 ]
squared_evens
```

```python
# What will be its output ?
numbers = [1, 2, 3, 4, 5]

while numbers:
    val = numbers.pop(0)
    if val % 2 == 0:
        continue
    print(val, end=" ")
```
