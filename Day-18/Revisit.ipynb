### PassCheck Class

pass_check = PassCheck("hello23456")

pass_check() -> Strong or Weak Password

Strong:
* length >= 8
* alpha + numerical

```python
class PassCheck:
    def __init__(self, password: str):
        """ Class to check if password is strong or weak

        Parameters
        ----------
        password : str
            User Input, password string to check
        """
        self.password = password

    def __str__(self):
        return f"{self.password[0]}***{self.password[-1]}"

    def __call__(self):
        if len(self.password) >= 8 and not self.password.isalpha() :
            return "Strong Password"
        
        return "Weak Password"
    
```

```python
pass_check = PassCheck("hellodgaergarg")

pass_check()
```

```python
test_numbers = [
    -10, -1, 0, 1, 2, 3, 4, 5, 6, 7, 9, 11, 13, 15, 17, 19, 21, 25, 4294967297,
    27, 47, 49, 79, 97, 100, 523, 527, 1000, 7919, 10000, 104729, 1000000,
    999999999989, 1000036000099, 999999999999989, 100000000000000, 9999991
]

# filter prime numbers from the given list 
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = list(filter(is_prime, test_numbers))

print(prime_numbers)
```

```python
prime = [n for n in test_numbers if is_prime(n)]
```
