```python
## Relation between Parent Approval and Parent Relationship

table = pd.crosstab(student_df.parent_approve, student_df.relation_parent)
table = table[["Distant", "Fair", "Close", "Very close"]]
table.T
```

```python
table.T.plot(
    kind = "line",
    title = "Relation between Parent Approval and Parent Relationship",
    ylabel = "Student Count",
    xlabel = "Degree of Parent Relationship"
)
```

```python
## Relation between Parent Relationship and Grade in Highschool

grounp_parent_relation = student_df.groupby(by = "relation_parent")
```

```python
relation = grounp_parent_relation["highschool_grade"].aggregate(["min", "max", "mean"]).T
```

```python
relation
```

```python
relation = relation[["Distant", "Fair", "Close", "Very close"]]
```

```python
relation.T.plot(kind = "bar")
```

As shown in the chart, the minimum values appear to have an inverse relationship with **Parent Relation**, whereas the mean score for the **Distant** relationship appears to be higher than those of the other categories.
