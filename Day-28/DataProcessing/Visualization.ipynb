### 1. Bar Chart

```python
ax = titanic_df.Survived.value_counts().plot(
    kind = "bar",
    title = "Count of Survived Passangers",
    figsize = (4,4),
    legend = False,
    xlabel = "Survived (0 = No, 1 = Yes)",
    ylabel = "No. of Passangers",
    grid = False,
    yticks = range(0, 551, 50),
    colormap = "Accent"
    # color=["red", "green"]
)
# Add numbers above each bar
ax.bar_label(ax.containers[0], padding=2)
```

```python
titanic_df.value_counts(subset = ["Sex", "Survived"]).plot(
    kind = "bar",
    title = "Count of Survived Passangers",
    figsize = (4,4),
    legend = True,
    xlabel = "Survived (0 = No, 1 = Yes)",
    ylabel = "No. of Passangers",
    grid = False,
    yticks = range(0, 551, 50),
    colormap = "Accent"
    # color=["red", "green"]
)
```

```python
crosstab = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Pclass, 
)
crosstab
```

```python
crosstab.plot(
    kind = "bar",
    title = "Count of Survived Passangers",
    figsize = (4,4),
    legend = True,
    xlabel = "Survived (0 = No, 1 = Yes)",
    ylabel = "No. of Passangers",
    grid = False,
    yticks = range(0, 501, 50),
    colormap = "tab20"
    # color=["red", "green"]
)
```

### 2. Pie Chart

```python
group_by_embarked = titanic_df.groupby(by = "Embarked")
group_by_embarked.count()
```

```python
group_by_embarked.count().plot(
    kind = "pie",
    y = "Survived",
    title = "Count of Passangers according to Pclass",
    autopct = "%1.1f%%",
    startangle = 90,
    figsize = (4,4)
)
```

```python
crosstab = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex, 
)
crosstab
```

```python
crosstab.plot(
    kind = "pie",
    subplots = True,
    figsize = (5,5),
)
```

### 3. Histogram

```python
titanic_df.Age.plot(
    kind = "hist",
    bins = 20,
    grid = True,
    figsize = (12,4)
)

titanic_df.Age.mean()
```

```python
titanic_df.Age.plot(
    kind = "kde",
    grid = True,
    ind = range(0,81),
    figsize = (12,4)
)
```

```python

titanic_df[["Age", "Survived"]].plot(
    kind = "hist",
    by = "Survived",
    bins = 20,
    figsize = (12,5)
)
```

```python

titanic_df[["Age", "Fare"]].plot(
    kind = "kde",
    figsize = (12,3),
    ind = range(0,200)
)
```

### 4. Line Graph

```python
titanic_df.Fare.loc[:100].plot(
    kind = "line",
    figsize = (15,2)
)
```

```python
titanic_df[["Fare","Age"]].loc[:100].plot(
    kind = "line",
    figsize = (15,2),
    style=["--", "-"]
)
```

### 5. Scatter Plot

```python
titanic_df.plot(
    kind = "scatter",
    x = "Age", y = "Fare",
    colormap = "bwr"
)
```
