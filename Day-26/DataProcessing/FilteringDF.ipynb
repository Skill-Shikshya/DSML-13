## Filtering using existing methods

````python
# isna()
null_cabin = titanic_df[titanic_df.Cabin.isna()]
````

````python
null_cabin.head()
````

````python
null_cabin.shape
````

````python
# Total Null values
titanic_df.Cabin.isna().sum()
````

````python
# notna()
non_empty_cabin = titanic_df[ titanic_df.Cabin.notna()]
non_empty_cabin.shape
````

````python
# duplicated data
duplicate_ticket = titanic_df[titanic_df.duplicated(subset=["Ticket"], keep = "first")] 
duplicate_ticket.head()
````

````python
titanic_df[titanic_df.Ticket == "349909"]
````

````python
# Duplicated row
titanic_df[titanic_df.duplicated()]
````

````python
# isin
titanic_df[titanic_df.Ticket.isin(["349909"])]
````

````python
# where
# mask

titanic_df.Age.where( titanic_df.Age > 0 , other= 1)
# titanic_df.Age.mask( titanic_df.Age < 0 , other= 1)
````
