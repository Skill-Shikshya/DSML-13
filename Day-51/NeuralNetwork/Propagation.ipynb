### Forward Propogation



<h1><b>Backpropagation</b></h1>

Backpropagation algorithm is an algorithm that can be used to train many kinds of neural network architecture. Backpropagation is named so as there is a backward propagation of errors in order to adapt the weights or parameters in the neural network.



Before diving deep into backpropagation let's understand concept of **chain rule** and **local gradient**.

In calculus, the chain rule is a formula to compute the derivative of a composite function. Let's assume we have an input $x$ and $y=f(x)$ and $z=g(y)$.

$\frac{\partial{z}}{\partial{x}}$ is change in value of $z$ with change in $x$ value. Since $z$ depend on $y$ and $y$ depend on $x$, $z$ is linked with $x$ through an intermediate variable $y$. In this case, the chain rule states that:$$\frac{\partial{z}}{\partial{x}}=\frac{\partial{z}}{\partial{y}}.\frac{\partial{y}}{\partial{x}}$$

Suppose, we have $x$ and $y$ as input which is feed to $z$, where $z=f(x,y)$ and $z$ is finally feed to $L$,where $L=g(z)$. Then , local gradient for a node  can be defined as partial derivative of immediate output of that node with respect to it's immediate inputs.The figure below visually explains local gradient,



<p></p>

Here, the local gradients for first node are $\frac{\partial{Z}}{\partial{x}}$ and $\frac{\partial{Z}}{\partial{y}}$ and for the second node the local gradient is $\frac{\partial{L}}{\partial{Z}}$

Now, we have to find the local gradients $\frac{\partial{L}}{\partial{\hat{y}}}$,$\frac{\partial{\hat{y}}}{\partial{z}}$,$\frac{\partial{z}}{\partial{w_0}}$,  $\frac{\partial{z}}{\partial{w_1}}$ and  $\frac{\partial{z}}{\partial{w_2}}$

$$\frac{\partial{L}}{\partial{\hat{y}}}=\frac{\partial{-(y\log{\hat{y}}+(1-y)\log{(1-\hat{y})})}}{\partial{\hat{y}}}$$
$$\frac{\partial{L}}{\partial{\hat{y}}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})$$

for $\frac{\partial{\hat{y}}}{\partial{z}}$, we know $\hat{y}=\frac{1}{1+e^{-z}}$
$$\frac{\partial{\hat{y}}}{\partial{z}}=\frac{\partial({\frac{1}{1+e^{-z}}})}{\partial{z}}$$
$$\frac{\partial{\hat{y}}}{\partial{z}}=\frac{(1+e^{-z})*\frac{\partial{(1)}}{\partial{z}}-\frac{\partial{(1+e^{-z})}}{\partial{z}}*1}{(1+e^{-z})^2}$$

$$\frac{\partial{\hat{y}}}{\partial{z}}=\frac{0-(-e^{-z})}{(1+e^{-z})^2}$$

$$\frac{\partial{\hat{y}}}{\partial{z}}=\frac{e^{-z}}{(1+e^{-z})^2}$$

$$\frac{\partial{\hat{y}}}{\partial{z}}=\frac{1}{(1+e^{-z})}*\frac{e^{-z}}{(1+e^{-z})}$$

$$\frac{\partial{\hat{y}}}{\partial{z}}=\hat{y}*(1-\frac{1}{(1+e^{-z})})$$

$$\frac{\partial{\hat{y}}}{\partial{z}}=\hat{y}*(1-\hat{y})$$

for $\frac{\partial{z}}{\partial{w_0}}$,
$$\frac{\partial{z}}{\partial{w_0}}=\frac{\partial{(w_0+x_1*w_1+x_2*w_2)}}{\partial{w_0}}$$

$$\frac{\partial{z}}{\partial{w_0}}=1$$

Similarly, $\frac{\partial{z}}{\partial{w_1}}=x_1$ and $\frac{\partial{z}}{\partial{w_2}}=x_2$

Finally, using chain rule as described above, we get
$$\therefore\frac{\partial{L}}{\partial{\hat{y}}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})$$

$$\therefore\frac{\partial{L}}{\partial{z}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}*(1-\hat{y})$$


$$\therefore\frac{\partial{L}}{\partial{w_0}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w_0}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}(1-\hat{y})*1$$

$$\therefore\frac{\partial{L}}{\partial{w_1}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w_1}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}(1-\hat{y})*x_1$$

$$\therefore \frac{\partial{L}}{\partial{w_2}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w_2}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}(1-\hat{y})*x_2$$








## Weight update

Our main objective of backpropagation is to update learnable parameters of the network, i.e the weights.

Now, we update the weights as follows:

$new \space w = old \space w + \Delta w = old \space w + \eta (- \frac{\partial L}{\partial w})$ 

where, $\eta$ is the learning rate.

For example,

$$new \space w_0 = old \space w_0 + \Delta w_0 = old \space w_0 + \eta (- \frac{\partial L}{\partial w_0})$$
$$new \space w_1 = old \space w_1 + \Delta w_1 = old \space w_1 + \eta (- \frac{\partial L}{\partial w_1})$$

$$new \space w_2 = old \space w_2 + \Delta w_2 = old \space w_2 + \eta (- \frac{\partial L}{\partial w_2})$$

Here, $\frac{\partial L}{\partial a^{(2)}_{1}}$ and $\frac{\partial L}{\partial a^{(2)}_{2}}$ are local gradients at node $L$.

*Above equation can be written in vector as,*


$$\frac{\partial L}{\partial \mathbf{(a^2)} }=-\mathbf{y}\left[\frac{1}{\mathbf{(a^2)}}\right]$$

### Network Architecture

Let us consider a logistic regression with two inputs variable $x_1$ and $x_2$. In graph below $z$ is a weighted sum of input plus bias. $\hat{y}$ is the output of the sogmoid function and $y$ is the true label for training example. $y$ and $\hat{y}$ are then passed to a binary cross entropy loss function to calculate the loss.



<p></p>

We have,

$$z = 1 * w_0 + x_1 * w_1 + x_2 * w_2$$

$$\hat{y} = \frac{1}{1 + e^{-(1 * w_0 + x_1 * w_1 + x_2 * w_2)}}$$

$$L = loss = - ( y \log{\hat{y}} + (1 - y) \log{(1-\hat{y})})$$





```python
# hyperparams
lr = 0.001
```

```python
# Forward Propogation

def linear(X, w):
    return np.matmul(X, w)

sigmoid = lambda z : 1 / ( 1+ np.exp(-z))

def calc_loss(y_hat, y):
    loss = 0
    for actual, pred in zip(y, y_hat):
        loss += -(actual * np.log(pred) + (1-actual) * np.log(1- pred))

    mean_loss = loss[0] / len(y)
    return float(mean_loss)
```

```python
X_with_bias = np.concat(
    ( np.ones((X.shape[0], 1)), X ),
    axis = 1
)

w = np.random.rand(3,1)
```

```python
z = linear(X_with_bias, w)
```

```python
y_hat = sigmoid(z)
```

```python
loss = calc_loss(y_hat, y)
```

**Backpropagation**

Backpropagation and chainrule seems as below,

<p></p>


from above reading material,we have derived formulas , let's state it and apply in code



$$\frac{\partial{L}}{\partial{w_0}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w_0}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}(1-\hat{y})*1$$

$$\frac{\partial{L}}{\partial{w_1}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w_1}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}(1-\hat{y})*x_1$$

$$ \frac{\partial{L}}{\partial{w_2}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w_2}}=-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})*\hat{y}(1-\hat{y})*x_2$$


**Vectorization**
$$\frac{\partial{L}}{\partial{w}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}}*\frac{\partial{z}}{\partial{w}}=\frac{\partial{L}}{\partial{z}}*\frac{\partial{z}}{\partial{w}}={x}^\intercal*(-(\frac{y}{\hat{y}}-\frac{(1-y)}{(1-\hat{y})})) \circ \hat{y}(1-\hat{y})$$



Here,
 for 200 training example $ \frac{\partial{L}}{\partial{\hat{y}}}$ will be of shape $200\times1$ and $ \frac{\partial{\hat{y}}}{\partial{z}}$ will also be of shape $200\times1$

$$ \frac{\partial{L}}{\partial{z}}=\frac{\partial{L}}{\partial{\hat{y}}}*\frac{\partial{\hat{y}}}{\partial{z}} $$

Since, there is one-to-one mapping between $z$, $\hat{y}$ and $L$ we compute elementwise multipication between $ \frac{\partial{L}}{\partial{\hat{y}}}$ and $ \frac{\partial{\hat{y}}}{\partial{z}}$, so the shape of $ \frac{\partial{L}}{\partial{z}}$ is $200\times1$. $x^\intercal$ is of the shape $3\times200$  so finally matrix multipication of $x^\intercal$ and $ \frac{\partial{L}}{\partial{z}}$ we gat $ \frac{\partial{L}}{\partial{w}}$ of shape $3\times1$.

 



```python
# BAck Propogation

def backpropogation(x, y, epochs, w):
    history = []

    for i in range(epochs):
        # forward - > loss
        y_hat = sigmoid(linear(x, w))
        history.append(calc_loss(y_hat, y))

        # Backward
        dl_dyhat = -(y / y_hat - (1-y) / (1- y_hat))
        dyhat_dx = y_hat * (1- y_hat)

        dl_dw = np.matmul(x.T , (dl_dyhat * dyhat_dx))

        w = w - (lr * dl_dw )

        print(f"Loss at epoch {i} is {calc_loss(y_hat, y):.4f}")

    return w, history
        
```

```python
y = y.reshape(-1,1)
```

```python
model, history = backpropogation( X_with_bias, y, 50, w)
```

```python
plt.plot(history)
```

```python
model
```

```python
# Inference
y_hat = sigmoid(linear(X_with_bias, model))
```

```python
pred = y_hat > 0.5
preds = pred.astype(int)
preds.flatten()
```
