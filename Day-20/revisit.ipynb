```python
# Factorial 
# 10! -> 9! * 10 -> 8! * 9 * 10
```

```python
cache = {0: 1, 1: 1}

def factorial(n):
    # Base case
    if n in cache.keys():
        return cache[n]
    # Recursive Case
    else:
        res = factorial(n-1) * n
        cache[n] = res
        return res

factorial(5)
```
